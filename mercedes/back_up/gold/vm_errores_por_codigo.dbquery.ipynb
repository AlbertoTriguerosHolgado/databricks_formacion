```sql
CREATE MATERIALIZED VIEW mercedes.gold.errores_por_codigo AS
SELECT
    message_id,
    MAX(message_text_en) AS descripcion_error,
    COUNT(*) AS total_errores
FROM mercedes.silver.mensajes
WHERE transactiontype = 'MESSAGES'
  AND subtype = 'TE1500AC'
  AND message_id IS NOT NULL
GROUP BY message_id;
```
