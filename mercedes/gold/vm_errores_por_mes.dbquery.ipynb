```sql
CREATE MATERIALIZED VIEW mercedes.gold.errores_por_mes AS
SELECT
    CAST(DATE_TRUNC('MONTH', timestamp_mensaje) AS DATE) AS mes,
    COUNT(*) AS total_errores
FROM mercedes.silver.mensajes
WHERE transactiontype = 'MESSAGES'
  AND subtype = 'TE1500AC'
  AND timestamp_mensaje IS NOT NULL
GROUP BY
    CAST(DATE_TRUNC('MONTH', timestamp_mensaje) AS DATE);
```
