```sql
CREATE MATERIALIZED VIEW mercedes.gold.errores_por_tool AS
SELECT
    tool,
    COUNT(*) AS total_errores,
    SUM(
        CASE WHEN in_process = TRUE THEN 1 ELSE 0 END
    ) AS errores_en_proceso,
    SUM(
        CASE WHEN in_process = FALSE THEN 1 ELSE 0 END
    ) AS errores_fuera_proceso,
    SUM(
        CASE WHEN in_process IS NULL THEN 1 ELSE 0 END
    ) AS errores_sin_clasificar
FROM mercedes.silver.mensajes
WHERE transactiontype = 'MESSAGES'
  AND subtype = 'TE1500AC'
  AND tool IS NOT NULL
GROUP BY tool;
```
